CELL 1: Import Libraries

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers, models
import matplotlib.pyplot as plt
import numpy as np
from sklearn.metrics import confusion_matrix, classification_report
```

CELL 2: Load and Preprocess Data

```python
(x_train, y_train), (x_test, y_test) = keras.datasets.mnist.load_data()  # Load MNIST dataset
x_train = x_train / 255.0  # Normalize training images to [0,1]
x_test = x_test / 255.0  # Normalize test images to [0,1]
x_train = x_train.reshape(-1, 28, 28, 1)  # Add channel dimension (28x28x1)
x_test = x_test.reshape(-1, 28, 28, 1)  # Add channel dimension for test data
print(f"Training samples: {x_train.shape[0]}, Test samples: {x_test.shape[0]}")
```

CELL 3: Visualize Sample Image

```python
plt.figure(figsize=(10,5))  # Create figure for visualization
for i in range(10):  # Show first 10 images
    plt.subplot(2, 5, i+1)  # Create 2x5 grid
    plt.imshow(x_train[i].reshape(28,28), cmap='gray')  # Display image
    plt.title(f"Label: {y_train[i]}")  # Show label
    plt.axis('off')  # Hide axes
plt.tight_layout()
plt.show()
```

CELL 4: Build CNN Model


```python
model = models.Sequential([
    layers.Conv2D(32, (3,3), activation='relu', input_shape=(28,28,1)),  # First conv layer: 32 filters
    layers.MaxPooling2D((2,2)),  # Reduce spatial dimensions by half
    layers.Conv2D(64, (3,3), activation='relu'),  # Second conv layer: 64 filters
    layers.MaxPooling2D((2,2)),  # Second pooling layer
    layers.Flatten(),  # Convert 2D features to 1D
    layers.Dense(64, activation='relu'),  # Fully connected layer with 64 neurons
    layers.Dense(10, activation='softmax')  # Output layer: 10 classes (digits 0-9)
])
```

CELL 5: Display Model Architecture

```python
model.summary()  # Show detailed layer-by-layer structure
```

CELL 6: Compile Model

```python
model.compile(
    optimizer='adam',  # Adaptive learning rate optimizer
    loss='sparse_categorical_crossentropy',  # Loss function for multi-class classification
    metrics=['accuracy']  # Track accuracy during training
)
```

CELL 7: Train Model

```python
history = model.fit(
    x_train, y_train,  # Training data
    epochs=5,  # Number of complete passes through dataset
    validation_split=0.1,  # Use 10% of training data for validation
    verbose=1  # Show training progress
)
```

CELL 8: Plot Training History

```python
plt.figure(figsize=(12,4))  # Create figure with two subplots
plt.subplot(1,2,1)  # First subplot for accuracy
plt.plot(history.history['accuracy'], label='Train')  # Training accuracy
plt.plot(history.history['val_accuracy'], label='Validation')  # Validation accuracy
plt.title('Model Accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()
plt.grid(True)

plt.subplot(1,2,2)  # Second subplot for loss
plt.plot(history.history['loss'], label='Train')  # Training loss
plt.plot(history.history['val_loss'], label='Validation')  # Validation loss
plt.title('Model Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

CELL 9: Evaluate Model

```python
test_loss, test_acc = model.evaluate(x_test, y_test, verbose=0)  # Test on unseen data
print(f"\nTest Accuracy: {test_acc:.4f}")
print(f"Test Loss: {test_loss:.4f}")
```

CELL 10: Make Predictions

```python
predictions = model.predict(x_test, verbose=0)  # Get predictions for all test images
predicted_classes = np.argmax(predictions, axis=1)  # Convert probabilities to class labels
```

CELL 11: Visualize Sample Prediction

```python
sample = 0  # Index of sample to visualize
plt.figure(figsize=(10,4))
plt.subplot(1,2,1)  # Show image
plt.imshow(x_test[sample].reshape(28,28), cmap='gray')
plt.title(f"True: {y_test[sample]}, Predicted: {predicted_classes[sample]}")
plt.axis('off')

plt.subplot(1,2,2)  # Show confidence scores
plt.bar(range(10), predictions[sample])  # Bar chart of probabilities
plt.title('Prediction Confidence')
plt.xlabel('Digit Class')
plt.ylabel('Probability')
plt.xticks(range(10))
plt.tight_layout()
plt.show()
```

CELL 12: Confusion Matrix

```python
cm = confusion_matrix(y_test, predicted_classes)  # Create confusion matrix
plt.figure(figsize=(8,6))
plt.imshow(cm, cmap='Blues')  # Display as heatmap
plt.colorbar()  # Add color scale
plt.title('Confusion Matrix')
plt.xlabel('Predicted')
plt.ylabel('True')
plt.xticks(range(10))
plt.yticks(range(10))
for i in range(10):  # Add numbers to cells
    for j in range(10):
        plt.text(j, i, str(cm[i,j]), ha='center', va='center')
plt.show()
```

CELL 13: Classification Report

```python
print("\nClassification Report:")
print(classification_report(y_test, predicted_classes,
                          target_names=[str(i) for i in range(10)]))
```


# Machine Learning Terms

**Convolutional Neural Network (CNN)**: A special type of neural network designed for images. It learns to recognize patterns like edges, shapes, and eventually complex objects.

**MNIST Dataset**: A collection of 70,000 handwritten digit images (0-9) used for training and testing. Each image is 28x28 pixels in grayscale.

**Normalization**: Converting pixel values from 0-255 to 0-1 range. This helps the model learn faster and more stably.

**Convolutional Layer (Conv2D)**: A layer that slides small filters (kernels) over the image to detect features like edges, curves, and patterns. Each filter learns to recognize a specific pattern.

**Filters/Kernels**: Small matrices (like 3x3) that scan the image to extract features. More filters = more different patterns detected.

**ReLU Activation**: A function that keeps positive values and converts negative values to zero. It adds non-linearity so the network can learn complex patterns.

**Max Pooling**: Reduces image size by taking the maximum value in small regions (like 2x2). This makes the model focus on the most important features and reduces computation.

**Flatten**: Converts the 2D feature maps into a 1D array so it can be fed into regular neural network layers.

**Dense Layer**: A fully connected layer where every neuron connects to all neurons in the previous layer. Used for final decision making.

**Softmax**: Converts the final layer outputs into probabilities that sum to 1. Each value represents the confidence that the image belongs to that digit class.

**Optimizer (Adam)**: An algorithm that adjusts weights during training to minimize errors. Adam is popular because it adapts the learning speed automatically.

**Loss Function**: Measures how wrong the model's predictions are. Lower loss = better performance.

**Sparse Categorical Crossentropy**: A loss function used when we have multiple classes (0-9) and labels are integers (not one-hot encoded).

**Epoch**: One complete pass through the entire training dataset. More epochs = more learning opportunities.

**Validation Split**: A portion of training data (10%) set aside to check how well the model performs on unseen data during training.

**Batch Processing**: Processing multiple images at once rather than one at a time, which speeds up training.

**Confusion Matrix**: A table showing where the model made correct predictions (diagonal) and where it confused one digit for another.

**Accuracy**: Percentage of correct predictions out of total predictions.

**Precision**: Of all images predicted as a certain digit, how many were actually that digit.

**Recall**: Of all actual images of a certain digit, how many did the model correctly identify.

**F1-Score**: A balanced measure combining precision and recall into one metric.

**Feature Maps**: The outputs of convolutional layers - they highlight where specific features were detected in the image.

**Spatial Dimensions**: The width and height of the image. Pooling reduces these while keeping important information.
