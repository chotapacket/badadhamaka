Cell 1: Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import confusion_matrix, classification_report, accuracy_score
```

Cell 2: Load Iris Dataset

```python
# Load Iris dataset from URL (simpler than bill dataset)
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data"
colnames = ['sepal-length', 'sepal-width', 'petal-length', 'petal-width', 'Class']
irisdata = pd.read_csv(url, names=colnames)
print(irisdata.head())
print(f"Dataset shape: {irisdata.shape}")
```

Cell 3: Visualize Sample Data

```python
# Display unique classes in the dataset
print("\nClasses in dataset:")
print(irisdata['Class'].unique())
print(f"\nClass distribution:")
print(irisdata['Class'].value_counts())
```

Cell 4: Prepare Data

```python
# Separate features (X) and target (y)
X = irisdata.drop('Class', axis=1)  # Features: sepal & petal measurements
y = irisdata['Class']  # Target: 3 species of Iris
```

Cell 5: Split Data


```python
# Divide into train (80%) and test (20%)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)
print(f"Training samples: {len(X_train)}")
print(f"Testing samples: {len(X_test)}")
```

Cell 6: Train SVM Model

```python
# Create and train SVM classifier with linear kernel
svclassifier = SVC(kernel='linear')
svclassifier.fit(X_train, y_train)
print("Model trained successfully!")
```

Cell 7: Make Predictions

```python
# Predict on test data
y_pred = svclassifier.predict(X_test)
print("Predictions made!")
```

Cell 8: Evaluate Model

```python
# Display confusion matrix
print("Confusion Matrix:")
cm = confusion_matrix(y_test, y_pred)
print(cm)
# Display classification report
print("\nClassification Report:")
print(classification_report(y_test, y_pred))
# Display accuracy
print(f"\nAccuracy: {accuracy_score(y_test, y_pred):.4f}")
```

Cell 9: Visualize Confusion Matrix

```python
# Plot confusion matrix as heatmap
plt.figure(figsize=(8, 6))
plt.imshow(cm, cmap='Blues')
plt.title('Confusion Matrix - Iris Classification')
plt.xlabel('Predicted Class')
plt.ylabel('Actual Class')
plt.colorbar()
# Add text annotations
for i in range(3):
    for j in range(3):
        plt.text(j, i, cm[i, j], ha='center', va='center', color='red')
plt.show()
```

Cell 10: Test with Different Kernels

```python
# Compare Linear, RBF, and Polynomial kernels
kernels = ['linear', 'rbf', 'poly']
for kernel in kernels:
    svm = SVC(kernel=kernel)
    svm.fit(X_train, y_train)
    y_pred_k = svm.predict(X_test)
    acc = accuracy_score(y_test, y_pred_k)
    print(f"{kernel.upper()} Kernel Accuracy: {acc:.4f}")
```

## Machine Learning Terms

### Support Vector Machine (SVM)
- Finds the best line/boundary to separate different classes
- Like drawing a fence that best divides different groups
- Maximizes the gap (margin) between classes for better accuracy

### Hyperplane
- Decision boundary that separates classes
- In 2D: a line, in 3D: a plane, in higher dimensions: hyperplane
- Goal is to find the "best" hyperplane that maximizes margin

### Support Vectors
- Data points closest to the hyperplane
- Most important points that define the boundary
- Like the posts that hold up the fence

### Margin
- Distance from hyperplane to nearest data point on each side
- SVM tries to maximize this margin
- Larger margin = better generalization = better model

### Kernel
- Mathematical trick to transform data to higher dimensions
- Helps find linear boundary when data is not separable
- Like viewing data from different angle to make patterns clear

### Linear Kernel
- Draws straight line to separate classes
- Best for data that's already linearly separable
- Simplest and fastest kernel

### RBF Kernel (Radial Basis Function)
- Creates circular/rounded boundaries around data points
- Most flexible, works well for complex patterns
- Can overfit if not tuned properly

### Polynomial Kernel
- Creates curved boundaries using polynomial equations
- 'degree' parameter controls complexity of curve
- Higher degree = more complex boundary

### Multi-Class Classification
- Classifying into 3 or more classes
- Iris example: 3 species (setosa, versicolor, virginica)
- SVM uses "one-vs-rest" strategy internally

### Confusion Matrix
- Table showing correct and incorrect predictions
- Rows = actual class, Columns = predicted class
- Diagonal values = correct predictions (we want these high)

### Precision
- Out of predicted positives, how many were actually correct
- Precision = True Positives / (True Positives + False Positives)
- Important when false alarms are costly

### Recall (Sensitivity)
- Out of actual positives, how many were correctly found
- Recall = True Positives / (True Positives + False Negatives)
- Important when missing positives is costly

### F1-Score
- Harmonic mean of precision and recall
- Balances both metrics into single score
- Good overall performance indicator

### Accuracy
- Percentage of total correct predictions
- Accuracy = Correct Predictions / Total Predictions
- Simple metric but can mislead with imbalanced data

### Train-Test Split
- Dividing data into training (80%) and testing (20%)
- Train on one part, test on unseen part
- Ensures model doesn't just memorize training data

### Overfitting
- Model memorizes training data instead of learning patterns
- Performs well on training but poorly on new data
- Like cramming for exam without understanding

### Generalization
- Model's ability to perform well on new, unseen data
- The ultimate goal of machine learning
- Good generalization = learned real patterns, not noise
