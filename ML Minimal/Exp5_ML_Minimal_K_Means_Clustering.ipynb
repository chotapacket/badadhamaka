Cell 1: Import Libraries

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
```

Cell 2: Load Iris Dataset

```python
# Load the Iris flower dataset
iris = load_iris()
X = iris.data  # Features: sepal length, sepal width, petal length, petal width
y = iris.target  # Actual labels (not used in clustering)
print(f"Dataset shape: {X.shape}")
print(f"Features: {iris.feature_names}")
```

Cell 3: Apply K-Means Clustering

```python
# Create K-Means model with 3 clusters
kmeans = KMeans(n_clusters=3, init='k-means++', max_iter=300, random_state=0)
y_kmeans = kmeans.fit_predict(X)  # Fit and predict cluster labels
print(f"Cluster labels: {y_kmeans[:10]}")  # Show first 10 labels
```

Cell 4: Visualize Clusters (2D)

```python
# Plot clusters using first two features
plt.figure(figsize=(10, 6))
plt.scatter(X[y_kmeans == 0, 0], X[y_kmeans == 0, 1], s=100, c='red', label='Cluster 1')
plt.scatter(X[y_kmeans == 1, 0], X[y_kmeans == 1, 1], s=100, c='blue', label='Cluster 2')
plt.scatter(X[y_kmeans == 2, 0], X[y_kmeans == 2, 1], s=100, c='green', label='Cluster 3')
# Plot centroids
plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1],
            s=300, c='yellow', marker='*', edgecolors='black', label='Centroids')
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Sepal Width (cm)')
plt.title('K-Means Clustering of Iris Dataset')
plt.legend()
plt.show()
```

Cell 5: Elbow Method

```python
# Find optimal number of clusters using WCSS
wcss = []  # Within-Cluster Sum of Squares
for i in range(1, 10):
    kmeans_temp = KMeans(n_clusters=i, init='k-means++', max_iter=300, random_state=0)
    kmeans_temp.fit(X)
    wcss.append(kmeans_temp.inertia_)  # Store WCSS value
# Plot elbow curve
plt.figure(figsize=(8, 5))
plt.plot(range(1, 10), wcss, marker='o')
plt.xlabel('Number of Clusters')
plt.ylabel('WCSS')
plt.title('Elbow Method')
plt.grid()
plt.show()
```

Cell 6: Silhouette Score Method

```python
# Calculate silhouette score for different cluster numbers
SS = []
for i in range(2, 10):  # Silhouette needs at least 2 clusters
    kmeans_temp = KMeans(n_clusters=i, init='k-means++', max_iter=300, random_state=0)
    labels = kmeans_temp.fit_predict(X)
    ss = silhouette_score(X, labels, metric='euclidean')
    SS.append(ss)
# Plot silhouette scores
plt.figure(figsize=(8, 5))
plt.plot(range(2, 10), SS, marker='o', color='orange')
plt.xlabel('Number of Clusters')
plt.ylabel('Silhouette Score')
plt.title('Silhouette Score Method')
plt.grid()
plt.show()
```

Cell 7: Print Cluster Centers

```python
# Display the centroid coordinates
print("Cluster Centers:")
print(kmeans.cluster_centers_)
print(f"\nOptimal clusters (from elbow/silhouette): 3")
```

Cell 8: Compare with Actual Labels

```python
# Create comparison table
comparison = pd.DataFrame({
    'Actual Label': y,
    'Cluster Label': y_kmeans
})
print("\nFirst 10 samples comparison:")
print(comparison.head(10))
```

## Machine Learning Terms

### Clustering
- Grouping similar data points together without predefined labels
- Unsupervised learning (no "correct answers" provided)
- Like organizing books by similarity without knowing categories

### K-Means Clustering
- Algorithm that divides data into K groups (K is specified by user)
- Groups data based on feature similarity
- Iteratively assigns points to nearest cluster center

### K (Number of Clusters)
- User-defined parameter specifying how many groups to create
- Too small K = under-clustering, too large K = over-clustering
- Need methods to find optimal K

### Centroid
- Center point of a cluster (mean of all points in cluster)
- Represents the "average" of the cluster
- Like the center of gravity for a group of points

### Euclidean Distance
- Straight-line distance between two points
- Used to measure similarity between data points
- Formula: √[(x₂-x₁)² + (y₂-y₁)²] for 2D

### K-Means++ Initialization
- Smart way to choose initial cluster centers
- Better than random initialization
- Helps algorithm converge faster and avoid poor solutions

### WCSS (Within-Cluster Sum of Squares)
- Measures how tight/compact clusters are
- Sum of squared distances from points to their centroid
- Lower WCSS = better clustering (points closer to centers)

### Elbow Method
- Graph technique to find optimal number of clusters
- Plot WCSS vs number of clusters
- Look for "elbow" point where WCSS reduction slows down

### Silhouette Score
- Measures how well each point fits in its cluster
- Range: -1 to +1 (higher is better)
- +1 = perfect clustering, 0 = overlapping clusters, -1 = wrong clusters

### Inertia
- Another name for WCSS
- Built-in attribute in sklearn: kmeans.inertia_
- Measures cluster cohesion

### Unsupervised Learning
- Learning patterns from data without labels
- Algorithm discovers hidden structure on its own
- Unlike supervised learning (which needs labeled data)

### Feature Similarity
- How alike data points are based on their attributes
- K-Means uses distance to measure similarity
- Closer points = more similar

### Convergence
- When cluster centers stop moving between iterations
- Algorithm has found stable solution
- Stopping criteria for K-Means

### Iteration
- One complete pass through the algorithm
- Each iteration: assign points → update centroids
- Continues until convergence or max_iter reached

### Random State
- Seed for reproducibility of random operations
- Same random_state = same results every time
- Useful for debugging and comparing results

### Anomaly Detection
- Finding unusual data points that don't fit any cluster
- Points far from all centroids might be anomalies
- Can add outlier points to see if they form separate clusters

### Visualization
- Plotting clusters in 2D or 3D space
- Helps understand how algorithm grouped data
- Different colors for different clusters

### Trade-offs
- More clusters (higher K) = lower WCSS but risk overfitting
- Fewer clusters (lower K) = simpler but might miss patterns
- Balance between simplicity and accuracy
