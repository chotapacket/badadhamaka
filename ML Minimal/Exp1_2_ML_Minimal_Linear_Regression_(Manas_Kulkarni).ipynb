Cell 1: Import Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

Cell 2: Load Dataset

```python
# Read the Excel file containing input-output data
data = pd.read_excel('/content/Simpleio.xlsx')
# Display first few rows
print(data.head())
```

Cell 3: Prepare Data

```python
# Rename columns to 'X' and 'Y' based on their order
data.columns = ['X', 'Y']

# Separate input (X) and output (Y) variables
X = data[['X']]  # Input feature (must be 2D array)
Y = data['Y']    # Output target (1D array)
```

Cell 4: Split Data

```python
# Divide data into training (80%) and testing (20%) sets
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2)
```

Cell 5: Create and Train Model

```python
# Create a linear regression model (single neuron)
model = LinearRegression()
# Train the model using training data
model.fit(X_train, Y_train)
```

Cell 6: Make Predictions

```python
# Predict output values for test data
Y_pred = model.predict(X_test)
```

Cell 7: Visualize Results

```python
# Plot actual vs predicted values
plt.scatter(X_test, Y_test, color='blue', label='Actual')
plt.plot(X_test, Y_pred, color='red', label='Predicted Line')
plt.xlabel('Input (X)')
plt.ylabel('Output (Y)')
plt.title('Simple Linear Regression')
plt.legend()
plt.show()
```

Cell 8: Model Parameters

```python
# Display slope (weight) and intercept (bias)
print(f"Slope (Weight): {model.coef_[0]}")
print(f"Intercept (Bias): {model.intercept_}")
```

Cell 9: Error Analysis

```python
# Calculate prediction errors (residuals)
errors = Y_test - Y_pred
# Plot error distribution
plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
plt.scatter(Y_pred, errors, color='purple')
plt.axhline(y=0, color='black', linestyle='--')
plt.xlabel('Predicted Values')
plt.ylabel('Errors (Actual - Predicted)')
plt.title('Error Plot')
plt.subplot(1, 2, 2)
plt.hist(errors, bins=20, color='orange', edgecolor='black')
plt.xlabel('Error')
plt.ylabel('Frequency')
plt.title('Error Distribution')
plt.tight_layout()
plt.show()
```

#Cell 10: ML Terms Explained

LINEAR REGRESSION: Finding a straight line that best fits the data points
- Like drawing a line through scattered dots on paper

TRAINING: Teaching the model using known data
- Showing examples so it learns the pattern

TESTING: Checking if the model works on new unseen data
- Like a surprise quiz after studying

FEATURES (X): Input values we use to make predictions
- What we know (like years of experience)

TARGET (Y): Output values we want to predict
- What we want to find out (like salary)

MODEL: The mathematical formula that connects input to output
- The "brain" that learned the pattern

SLOPE/WEIGHT: How steep the line is
- How much Y changes when X increases by 1

INTERCEPT/BIAS: Where the line crosses the Y-axis
- The starting value when X = 0

PREDICTION: Using the trained model to estimate Y for new X values
- Making educated guesses based on what we learned

FIT: Process of adjusting the line to match training data
- Fine-tuning the model to get best results

ERROR/RESIDUAL: Difference between actual and predicted values
- How far off our predictions are from reality

ERROR PLOT: Graph showing prediction mistakes
- Helps us see if model makes consistent or random errors

DISTRIBUTION: How errors are spread out
- Shows if errors are mostly small or some are very large
