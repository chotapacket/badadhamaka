CELL 1: Import Libraries

```python
import numpy as np
import matplotlib.pyplot as plt
```

CELL 2: XOR Dataset

```python
X = np.array([[0,0], [0,1], [1,0], [1,1]])  # XOR inputs
y = np.array([[0], [1], [1], [0]])  # XOR outputs
```

CELL 3: Initialize Weights

```python
np.random.seed(1)  # For reproducible results
W1 = np.random.randn(2, 2)  # Input to hidden layer weights (2x2)
b1 = np.zeros((1, 2))  # Hidden layer bias
W2 = np.random.randn(2, 1)  # Hidden to output layer weights (2x1)
b2 = np.zeros((1, 1))  # Output layer bias
```

CELL 4: Activation Functions

```python
def tanh(x):  # Activation for hidden layer
    return np.tanh(x)

def tanh_derivative(x):  # Derivative of tanh
    return 1 - np.tanh(x)**2

def sigmoid(x):  # Activation for output layer
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative(x):  # Derivative of sigmoid
    return x * (1 - x)
```

CELL 5: Training Loop

```python
lr = 0.5  # Learning rate
epochs = 10000  # Number of training iterations
losses = []  # To track error over time

for i in range(epochs):
    # Forward Pass
    z1 = np.dot(X, W1) + b1  # Hidden layer input
    a1 = tanh(z1)  # Hidden layer output
    z2 = np.dot(a1, W2) + b2  # Output layer input
    a2 = sigmoid(z2)  # Final output

    # Compute Loss (Mean Squared Error)
    loss = np.mean((y - a2)**2)
    losses.append(loss)

    # Backward Pass (Backpropagation)
    d2 = (a2 - y) * sigmoid_derivative(a2)  # Output layer error
    dW2 = np.dot(a1.T, d2)  # Gradient for W2
    db2 = np.mean(d2, axis=0, keepdims=True)  # Gradient for b2

    d1 = np.dot(d2, W2.T) * tanh_derivative(z1)  # Hidden layer error
    dW1 = np.dot(X.T, d1)  # Gradient for W1
    db1 = np.mean(d1, axis=0, keepdims=True)  # Gradient for b1

    # Update Weights
    W2 -= lr * dW2  # Update output weights
    b2 -= lr * db2  # Update output bias
    W1 -= lr * dW1  # Update hidden weights
    b1 -= lr * db1  # Update hidden bias
```

CELL 6: Test Network

```python
print("XOR Predictions:")
z1 = np.dot(X, W1) + b1  # Forward pass through hidden layer
a1 = tanh(z1)
z2 = np.dot(a1, W2) + b2  # Forward pass through output layer
predictions = sigmoid(z2)

for i in range(4):
    print(f"Input: {X[i]} -> Predicted: {predictions[i][0]:.4f}, Rounded: {round(predictions[i][0])}")
```

CELL 7: Plot Training Loss

```python
plt.plot(losses)  # Plot loss over epochs
plt.xlabel('Epoch')  # X-axis label
plt.ylabel('Loss (MSE)')  # Y-axis label
plt.title('Training Loss Over Time')  # Chart title
plt.yscale('log')  # Log scale for better visualization
plt.grid(True)  # Add grid
plt.show()
```

CELL 8: Explanation of ML Terms

# Machine Learning Terms
**Neural Network**: A system that mimics how our brain works, with layers of connected nodes that process information.

**XOR Problem**: A classic puzzle where output is 1 only when inputs are different (0,1 or 1,0), and 0 when inputs are same.

**Weights (W1, W2)**: Numbers that determine how important each connection is, like turning a volume knob up or down.

**Bias (b1, b2)**: A constant added to help the network make better predictions, like an adjustment factor.

**Activation Function**: A mathematical function that decides if a neuron should "fire" or not:
  - **tanh**: Squashes values between -1 and 1, used in hidden layer
  - **sigmoid**: Squashes values between 0 and 1, used for final output

**Forward Pass**: Information flows from input → hidden layer → output, making a prediction.

**Loss/Error**: How wrong our prediction is compared to the actual answer. We want this to be as small as possible.

**Mean Squared Error (MSE)**: Averages the squared differences between predictions and actual values.

**Backpropagation**: The learning process where we calculate how much each weight contributed to the error and adjust them accordingly.

**Gradient**: Shows the direction to adjust weights to reduce error (like following a slope downhill).

**Learning Rate (lr)**: Controls how big the weight adjustments are (small steps = slow but steady learning).

**Epoch**: One complete pass through all training data. More epochs = more learning opportunities.

**Derivative**: Measures how much the output changes when we slightly change the input (used to calculate gradients).
