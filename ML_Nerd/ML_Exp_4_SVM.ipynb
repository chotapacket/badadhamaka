```python
import pandas as pd                         # Import Pandas library
import numpy as np                          # Import NumPy library
import matplotlib.pyplot as plt             # Import Matplotlib library
%matplotlib inline
from google.colab import files
uploaded = files.upload()
```

```python
import io
bankdata = pd.read_csv(io.BytesIO(uploaded["bill_authentication.csv"]))  # Reads uploaded excel File
bankdata                                                                 # Displays the loaded data in notebook
```

```python
bankdata.shape                # Displays number of rows and columns
```

```python
bankdata.head()               # Displays first 5 rows
```

```python
# Split the data into attributes/features and labels/target
X = bankdata.drop('Class', axis=1)                            # Drop Class from dataframe
y = bankdata['Class']
y
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.25)         # 75% training and 25% testing
X_train
```

```python
from sklearn.svm import SVC                  # Imports the Support Vector Classification class from scikit-learn
svclassifier = SVC(kernel='linear')          # Use linear kernel and separate the data
svclassifier.fit(X_train, y_train)
```

```python
#To make predictions
y_pred = svclassifier.predict(X_test)
```

Evaluating the Algorithm

Confusion matrix, precision, recall, and F1 measures are the most commonly used metrics for classification tasks. Scikit-Learn's metrics library contains the classification_report and confusion_matrix methods, which can be readily used to find out the values for these important metrics.

```python
from sklearn.metrics import classification_report, confusion_matrix
print(confusion_matrix(y_test,y_pred))                                  # Displays confusion matrix and classification report
print(classification_report(y_test,y_pred))
```

```python
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data"

# Assign colum names to the dataset
colnames = ['sepal-length', 'sepal-width', 'petal-length', 'petal-width', 'Class']
# Read dataset to pandas dataframe
irisdata = pd.read_csv(url, names=colnames)
```

```python
irisdata
```

```python
X = irisdata.drop('Class', axis=1)
y = irisdata['Class']
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.25)        # 75% training and 25% testing
X_train
```

```python
from sklearn.svm import SVC
# svclassifier = SVC(kernel='poly', degree=8)
svclassifier = SVC(kernel='rbf') #Gaussian Kernel
# svclassifier = SVC(kernel='sigmoid') #Sigmoid Kernel
svclassifier.fit(X_train, y_train)
```

```python
y_pred = svclassifier.predict(X_test)
```

```python
from sklearn.metrics import classification_report, confusion_matrix
print(confusion_matrix(y_test, y_pred))                                  # Displays confusion matrix and classification report
print(classification_report(y_test, y_pred))
```
