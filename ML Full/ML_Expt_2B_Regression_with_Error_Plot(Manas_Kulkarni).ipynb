```python
# Import necessary libraries for building and training a neural network, data manipulation, and plotting.
import tensorflow as tf
from tensorflow import keras
print(tf.__version__)
print(keras.__version__)
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# Import modules for splitting data into training and testing sets and for scaling features.
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
# Load the housing dataset from a CSV file into a pandas DataFrame.
housing=pd.read_csv('/content/housing.csv')
# Display the DataFrame.
housing
```

```python
# Remove the 'ocean_proximity' and 'total_bedrooms' columns from the DataFrame.
housing.pop('ocean_proximity')
housing.pop('total_bedrooms')
```

```python
# Normalize the housing data by scaling features to a range between 0 and 1.
hnorm=(((housing-housing.min())/(housing.max()-housing.min())))
# Check for any missing values in the normalized DataFrame.
hnorm.isna().sum()
```

```python
# Check the data types of the columns in the original housing DataFrame.
housing.dtypes
```

```python
# Access and display the 'total_rooms' column of the original housing DataFrame.
housing.total_rooms
```

```python
# Separate the target variable ('median_house_value') from the features and store it in 'target'.
target = housing.pop('median_house_value')
```

```python
# Convert the target variable and normalized features to NumPy arrays.
Y=target.values
X=hnorm.values
# Display the feature array.
X
```

```python
# Split the data into training and testing sets (85% training, 15% testing).
X_train, X_test, y_train, y_test = train_test_split(X, Y,test_size=0.15)
# Display the shape of the training target variable array.
y_train.shape
# Create a validation set from the training data (last 1544 samples).
x_val=X_train[16000:]
y_val=y_train[16000:]
# Reduce the training set to the first 16000 samples.
X_train=X_train[:16000]
y_train=y_train[:16000]
```

```python
# Define a sequential Keras model with a dense hidden layer and a dense output layer.
model = keras.models.Sequential([
# Add a dense layer with 128 units, ReLU activation, and input shape of 8.
keras.layers.Dense(128, activation="relu",input_shape=(8,)),
# (Optional) Add another dense layer with 16 units and ReLU activation. (Currently commented out)
#keras.layers.Dense(16, activation="relu"),
# Add a dense output layer with 1 unit (for regression).
keras.layers.Dense(1)
])
# Compile the model with Mean Absolute Error (MAE) loss and Stochastic Gradient Descent (SGD) optimizer.
model.compile(loss="mae", optimizer="sgd")
# Train the model for 50 epochs, using the training data and validating with the validation data.
history = model.fit(X_train, y_train, epochs=50,verbose=1,validation_data=(x_val, y_val))
```

```python
# Get the training history from the model fitting process.
history_dict = history.history
# Display the keys of the history dictionary (loss and validation loss).
history_dict.keys()
```

```python
# Get the training history dictionary.
history_dict = history.history
# Extract the loss values for training and validation.
loss_values = history_dict['loss']
val_loss_values = history_dict['val_loss']
# (Optional) Extract accuracy values if available. (Currently commented out)
#acc = history_dict['accuracy']
# Create a range of epochs for plotting.
epochs = range(1, len(loss_values)+1 )
# Create a figure for the plot.
plt.figure(figsize=(6, 4))
# (Optional) Create subplots if needed. (Currently commented out)
#plt.subplot(121)
# Plot the training loss in red.
plt.plot(epochs, loss_values, 'r', label='Training loss')
# Plot the validation loss in blue.
plt.plot(epochs, val_loss_values, 'b', label='Validation loss')
# Set the title of the plot.
plt.title('Training and validation loss')
# Set the x-axis label.
plt.xlabel('Epochs')
# Set the y-axis label.
plt.ylabel('Loss')
# Add a legend to the plot.
plt.legend()
# Display the plot.
plt.show()
```

```python
# Evaluate the trained model on the test data and store the loss in 'htest'.
htest = model.evaluate(X_test, y_test)
```

```python
# Select the first 50 samples from the test features as new instances for prediction.
X_new = X_test[:50] # pretend these are new instances
# Predict the target values for the new instances using the trained model.
y_pred = model.predict(X_new)
# Create a figure for the plot.
plt.figure(figsize=(10,5))
# Plot the actual target values for the first 50 test samples in blue dashed line.
plt.plot(np.arange(0,len(X_new),1),y_test[:50],'b--',label='Actual')
# Plot the predicted target values for the new instances in red dashed line.
plt.plot(np.arange(0,len(X_new),1),y_pred,'r--',label='Predicted')
# Add a legend to the plot.
plt.legend()
# Display the plot.
plt.show()
plt.show()
# (Optional) Scatter plot of actual vs. predicted values. (Currently commented out)
#plt.scatter(y_test[:20],y_pred)
```
