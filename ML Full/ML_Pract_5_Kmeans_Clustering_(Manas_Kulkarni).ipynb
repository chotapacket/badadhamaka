```python
import numpy as np # Imports the numpy library, commonly used for numerical operations
import matplotlib.pyplot as plt # Imports the matplotlib.pyplot library, used for creating plots and visualizations
from sklearn.datasets import load_iris # Imports the load_iris function from scikit-learn, which provides the Iris dataset
import pandas as pd # Imports the pandas library, used for data manipulation and analysis, particularly with DataFrames
from sklearn.cluster import KMeans # Imports the KMeans clustering algorithm from scikit-learn
```

```python
np.random.seed(25) # Sets the random seed for reproducibility of random number generation
X=np.random.randint(100,size=(100,2)) # Generates a 100x2 array of random integers between 0 and 99
print(X) # Prints the generated array X
```

```python
# Coverting program in anomoly detection
X[99]=[8,150] # Changes the value of the last row (index 99) in array X to [8, 150]
X[98]=[10,160] # Changes the value of the second to last row (index 98) in array X to [10, 160]
```

```python
print(X) # Prints the updated array X
```

```python
kmeans = KMeans(n_clusters = 5, init = 'k-means++', max_iter = 300, n_init = 10, random_state = 0) # Initializes the KMeans model with 5 clusters
y_kmeans = kmeans.fit_predict(X) # Fits the KMeans model to the data X and predicts the cluster for each data point
```

```python
print(y_kmeans) # Prints the cluster labels for each data point
```

```python
#Visualising the clusters
plt.scatter(X[y_kmeans == 0, 0], X[y_kmeans == 0, 1], s = 100, c = 'red') # Plots data points belonging to cluster 0 in red
plt.scatter(X[y_kmeans == 1, 0], X[y_kmeans == 1, 1], s = 100, c = 'blue') # Plots data points belonging to cluster 1 in blue
plt.scatter(X[y_kmeans == 2, 0], X[y_kmeans == 2, 1], s = 100, c = 'green') # Plots data points belonging to cluster 2 in green
plt.scatter(X[y_kmeans == 3, 0], X[y_kmeans == 3, 1], s = 100, c = 'cyan') # Plots data points belonging to cluster 3 in cyan
plt.scatter(X[y_kmeans == 4, 0], X[y_kmeans == 4, 1], s = 100, c = 'magenta') # Plots data points belonging to cluster 4 in magenta
#Plotting the centroids of the clusters
plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:,1], s = 200, c = 'yellow', label = 'Centroids') # Plots the cluster centroids in yellow

plt.legend() # Displays the legend for the plot
```

```python
from sklearn.cluster import KMeans # Imports the KMeans clustering algorithm
wcss = [] # Initializes an empty list to store the Within-Cluster Sum of Squares (WCSS)

for i in range(1, 10): # Loops through different numbers of clusters from 1 to 9
    kmeans = KMeans(n_clusters = i, init = 'k-means++', max_iter = 300, n_init = 10, random_state = 0) # Initializes KMeans for each number of clusters
    kmeans.fit(X) # Fits the KMeans model to the data
    wcss.append(kmeans.inertia_) # Appends the WCSS for the current number of clusters to the list

#Plotting the results onto a line graph, allowing us to observe 'The elbow'
plt.plot(range(1, 10), wcss) # Plots the WCSS against the number of clusters
plt.title('The elbow method') # Sets the title of the plot
plt.xlabel('Number of clusters') # Sets the label for the x-axis
plt.ylabel('WCSS') #within cluster sum of squares # Sets the label for the y-axis
plt.show() # Displays the plot
```

```python
from sklearn.metrics import silhouette_score # Imports the silhouette_score metric
SS=[] # Initializes an empty list to store silhouette scores
kmeans = KMeans(n_clusters = 5, init = 'k-means++', max_iter = 300, n_init = 10, random_state = 0) # Initializes KMeans with 5 clusters
kmeans.fit(X) # Fits the KMeans model to the data
labels = kmeans.labels_ # Gets the cluster labels for each data point
SS=silhouette_score(X,labels,metric='euclidean') # Calculates the silhouette score
print(SS) # Prints the silhouette score
print(labels) # Prints the cluster labels
```

```python
from sklearn.metrics import silhouette_score # Imports the silhouette_score metric
SS=[] # Initializes an empty list to store silhouette scores
for i in range(2,10): # Loops through different numbers of clusters from 2 to 9
    kmeans = KMeans(n_clusters = i, init = 'k-means++', max_iter = 300, n_init = 10, random_state = 0) # Initializes KMeans for each number of clusters
    kmeans.fit(X) # Fits the KMeans model to the data
    labels = kmeans.labels_ # Gets the cluster labels
    ss=silhouette_score(X,labels,metric='euclidean') # Calculates the silhouette score for the current number of clusters
    SS.append(ss) # Appends the silhouette score to the list

#Plotting the results onto a line graph, allowing us to observe 'The elbow'
plt.plot(range(2, 10),SS) # Plots the silhouette scores against the number of clusters
plt.title('The silhouette score method') # Sets the title of the plot
plt.xlabel('Number of clusters') # Sets the label for the x-axis
plt.ylabel('SS') # Sets the label for the y-axis
plt.show() # Displays the plot
```

**Iris Flower Clustering using K Means**

```python
iris = load_iris() # Loads the Iris dataset
X = iris.data # Gets the data (features) from the Iris dataset
y = iris.target # Gets the target (labels) from the Iris dataset
```

```python
kmeans = KMeans(n_clusters = 3, init = 'k-means++', max_iter = 300, n_init = 10, random_state = 0) # Initializes the KMeans model with 3 clusters
y_kmeans = kmeans.fit_predict(X) # Fits the KMeans model to the Iris data and predicts the cluster for each data point
```

```python
#Visualising the clusters
plt.scatter(X[y_kmeans == 0, 0], X[y_kmeans == 0, 1], s = 100, c = 'red', label = 'Iris-setosa') # Plots data points of cluster 0 (predicted as Iris-setosa) in red
plt.scatter(X[y_kmeans == 1, 0], X[y_kmeans == 1, 1], s = 100, c = 'blue', label = 'Iris-versicolour') # Plots data points of cluster 1 (predicted as Iris-versicolour) in blue
plt.scatter(X[y_kmeans == 2, 0], X[y_kmeans == 2, 1], s = 100, c = 'green', label = 'Iris-virginica') # Plots data points of cluster 2 (predicted as Iris-virginica) in green
#Plotting the centroids of the clusters
plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:,1], s = 200, c = 'yellow', label = 'Centroids') # Plots the cluster centroids in yellow

plt.legend() # Displays the legend
```

```python
from mpl_toolkits.mplot3d import Axes3D # Imports the Axes3D module for 3D plotting

fig = plt.figure(figsize=(8, 6)) # Creates a new figure
ax = fig.add_subplot(111, projection='3d') # Adds a 3D subplot to the figure

# Plot data points for each cluster
ax.scatter(X[y_kmeans == 0, 0], X[y_kmeans == 0, 1], X[y_kmeans == 0, 2], c='red', s=100, label='Iris-setosa') # Plots cluster 0 data in 3D
ax.scatter(X[y_kmeans == 1, 0], X[y_kmeans == 1, 1], X[y_kmeans == 1, 2], c='blue', s=100, label='Iris-versicolour') # Plots cluster 1 data in 3D
ax.scatter(X[y_kmeans == 2, 0], X[y_kmeans == 2, 1], X[y_kmeans == 2, 2], c='green', s=100, label='Iris-virginica') # Plots cluster 2 data in 3D

# Plot centroids
ax.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1], kmeans.cluster_centers_[:, 2], s=200, c='yellow', label='Centroids') # Plots centroids in 3D

ax.set_xlabel('Sepal Length (cm)') # Sets the label for the x-axis
ax.set_ylabel('Sepal Width (cm)') # Sets the label for the y-axis
ax.set_zlabel('Petal Length (cm)') # Sets the label for the z-axis
ax.set_title('3D K-Means Clustering of Iris Dataset') # Sets the title of the plot
ax.legend() # Displays the legend

plt.show() # Displays the 3D plot
```

# **Wine Dataset**

```python
from sklearn.datasets import load_wine # Imports the load_wine function from scikit-learn

wine = load_wine() # Loads the Wine dataset
df = pd.DataFrame(wine.data, columns=wine.feature_names) # Creates a pandas DataFrame from the wine data with feature names as columns
df['target'] = wine.target # Adds the target variable (wine type) as a new column to the DataFrame
display(df.head()) # Displays the first 5 rows of the DataFrame
display(df.info()) # Displays information about the DataFrame, including data types and non-null values
```

```python
from sklearn.preprocessing import StandardScaler # Imports the StandardScaler for feature scaling

scaler = StandardScaler() # Initializes the StandardScaler
X_scaled = scaler.fit_transform(df.drop('target', axis=1)) # Scales the features (all columns except 'target') of the DataFrame
```

```python
wcss = [] # Initializes an empty list to store WCSS values
for i in range(1, 11): # Loops through different numbers of clusters from 1 to 10
    kmeans = KMeans(n_clusters=i, init='k-means++', max_iter=300, n_init=10, random_state=0) # Initializes KMeans
    kmeans.fit(X_scaled) # Fits KMeans to the scaled data
    wcss.append(kmeans.inertia_) # Appends the WCSS to the list

plt.plot(range(1, 11), wcss) # Plots WCSS vs. number of clusters
plt.title('The elbow method') # Sets the title
plt.xlabel('Number of clusters') # Sets the x-label
plt.ylabel('WCSS') # Sets the y-label
plt.show() # Displays the plot

SS = [] # Initializes an empty list to store silhouette scores
for i in range(2, 11): # Loops through different numbers of clusters from 2 to 10
    kmeans = KMeans(n_clusters=i, init='k-means++', max_iter=300, n_init=10, random_state=0) # Initializes KMeans
    kmeans.fit(X_scaled) # Fits KMeans to the scaled data
    labels = kmeans.labels_ # Gets the cluster labels
    ss = silhouette_score(X_scaled, labels, metric='euclidean') # Calculates the silhouette score
    SS.append(ss) # Appends the silhouette score to the list

plt.plot(range(2, 11), SS) # Plots silhouette scores vs. number of clusters
plt.title('The silhouette score method') # Sets the title
plt.xlabel('Number of clusters') # Sets the x-label
plt.ylabel('SS') # Sets the y-label
plt.show() # Displays the plot
```

```python
kmeans_wine = KMeans(n_clusters = 3, init = 'k-means++', max_iter = 300, n_init = 10, random_state = 0) # Initializes KMeans for the wine dataset with 3 clusters
y_kmeans_wine = kmeans_wine.fit_predict(X_scaled) # Fits KMeans to the scaled wine data and predicts clusters
print(y_kmeans_wine) # Prints the cluster labels for the wine data
```

```python
# Visualize the clusters using the first two features of the scaled data
plt.figure(figsize=(8, 6)) # Creates a new figure with a specified size
plt.scatter(X_scaled[y_kmeans_wine == 0, 0], X_scaled[y_kmeans_wine == 0, 1], s = 100, c = 'red', label = 'Cluster 1: Lower values for most features (Wine Type 1)') # Plots cluster 0 (predicted as Wine Type 1) in red
plt.scatter(X_scaled[y_kmeans_wine == 1, 0], X_scaled[y_kmeans_wine == 1, 1], s = 100, c = 'blue', label = 'Cluster 2: Higher malic acid, ash, color intensity, lower hue (Wine Type 2)') # Plots cluster 1 (predicted as Wine Type 2) in blue
plt.scatter(X_scaled[y_kmeans_wine == 2, 0], X_scaled[y_kmeans_wine == 2, 1], s = 100, c = 'green', label = 'Cluster 3: Higher alcohol, phenols, flavanoids, proline, lower color intensity (Wine Type 0)') # Plots cluster 2 (predicted as Wine Type 0) in green

# Plot the centroids
plt.scatter(kmeans_wine.cluster_centers_[:, 0], kmeans_wine.cluster_centers_[:, 1], s = 200, c = 'yellow', label = 'Centroids', marker='*') # Plots the centroids in yellow with a star marker

plt.title('K-Means Clustering of Wine Dataset (Features 1 and 2)') # Sets the title of the plot
plt.xlabel(wine.feature_names[0] + ' (Scaled)') # Labeling the x-axis with the actual feature name
plt.ylabel(wine.feature_names[1] + ' (Scaled)') # Labeling the y-axis with the actual feature name
plt.legend() # Displays the legend
plt.show() # Displays the plot
```

Conclusion: K-Means clustering was applied to three different datasets: a synthetic dataset with anomalies, the Iris dataset, and the Wine dataset. For the synthetic dataset, the clustering visually identified the two introduced anomalies as separate points from the main clusters. For the Iris dataset, K-Means with 3 clusters effectively separated the three species, as shown by both the 2D and 3D visualizations. Finally, for the Wine dataset, the elbow method and silhouette score suggested that 3 clusters is a reasonable choice, and the clustering results, when visualized using the first two scaled features, show a clear separation of the data points into three groups, which aligns with the known three types of wine in the dataset.
